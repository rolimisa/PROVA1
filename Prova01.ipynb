A melhor estrutura de dados para armazenar os chamados aguardando atendimento é a fila, assim eles serão atendidos por ordem de chegada. FIFO 

A melhor estrutura para armazenar os chamados atendidos é a pilha, pois assim será possível retirar o ultimo que entrou - Lifo

Após executar a sequência: 

atender
atender
desfazer
atender

As estruturas ficarão da seguinte forma:

Aguardando:
Daniel
Bruno

Atendidos:
Ana
Bruno
Carla

```python
class CentralAtendimento:
    def __init__(self):
        self.aguardando = []
        self.atendidos = []

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = {
            'codigo': codigo,
            'cliente': cliente,
            'prioridade': prioridade
        }
        self.aguardando.append(chamado)

    def atender_proximo(self):
        if len(self.aguardando) == 0:
            print("Nenhum chamado na fila.")
            return None
        
        chamado = self.aguardando.pop(0)
        self.atendidos.append(chamado)
        return chamado

    def desfazer_ultimo_atendimento(self):
        if len(self.atendidos) == 0:
            print("Nenhum atendimento para desfazer.")
            return None
        
        chamado = self.atendidos.pop()
        self.aguardando.append(chamado)
        return chamado

    def buscar_chamado(self, codigo):
        for c in self.aguardando:
            if c['codigo'] == codigo:
                return c
        for c in self.atendidos:
            if c['codigo'] == codigo:
                return c
        return None

    def exibir_situacao(self):
        print("Aguardando ")
        for c in self.aguardando:
            print(c['codigo'], "-", c['cliente'], "-", c['prioridade'])
            
        print("Atendidos")
        for c in self.atendidos:
            print(c['codigo'], "-", c['cliente'], "-", c['prioridade'])
```

```python
central = CentralAtendimento()

central.adicionar_chamado(101, "Ana", "Média")
central.adicionar_chamado(102, "Bruno", "Alta")
central.adicionar_chamado(103, "Carla", "Baixa")
central.adicionar_chamado(104, "Daniel", "Alta")

central.atender_proximo()
central.atender_proximo()

central.desfazer_ultimo_atendimento()

central.atender_proximo()

central.exibir_situacao()

print(central.buscar_chamado(104))
```

No pior caso a complexidade é O (n), como a busca não utiliza índice, a complexidade de tempo cresce conforme o número de chamados aumenta.
